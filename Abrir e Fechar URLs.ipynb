```python
import webbrowser
import time
import os
import signal
import psutil  # instale com: pip install psutil

# Lista de URLs
urls = [
    "http://b.com",
    "http://b.com",
    "http://b.com",
    "http://b.com",
    "http://b.com",
    "http://b.com"

]

# Tempo em segundos que cada aba ficará aberta
tempo_aberto = 5

# Função para tentar fechar o navegador (opcional, depende do SO)
def fechar_navegador():
    for proc in psutil.process_iter(['pid', 'name']):
        if proc.info['name'] and (
            "chrome" in proc.info['name'].lower() or
            "firefox" in proc.info['name'].lower() or
            "msedge" in proc.info['name'].lower()
        ):
            try:
                proc.terminate()
            except Exception as e:
                print(f"Não foi possível encerrar {proc.info['name']}: {e}")

# Loop para abrir e fechar
for url in urls:
    print(f"Abrindo {url} ...")
    webbrowser.open(url)
    time.sleep(tempo_aberto)
    print(f"Fechando navegador...")
    fechar_navegador()
    time.sleep(2)

print("Todas as URLs foram abertas e fechadas com sucesso!")
```
